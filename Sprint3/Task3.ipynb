<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс BankAccount в C#, который будет представлять
информацию об учетных записях в банке. На основе этого класса разработать 2-3
производных класса, демонстрирующих принципы наследования и полиморфизма.
В каждом из классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.


Требования к базовому классу BankAccount:

• Атрибуты: Номер счета (AccountNumber), Баланс (Balance), Тип счета
(AccountType).


• Методы:
o GetInfo(): метод для получения информации о счете в виде строки.
o Deposit(decimal amount): метод для внесения денег на счет.
o Withdraw(decimal amount): метод для снятия денег со счета.

Требования к производным классам:

1. Сберегательный счет (SavingsAccount): Должен содержать дополнительные
атрибуты, такие как Процентная ставка (InterestRate). Метод Deposit() должен
быть переопределен для добавления процентов к сумме вклада при
внесении денег на счет.

2. Текущий счет (CheckingAccount): Должен содержать дополнительные
атрибуты, такие как Лимит овердрафта (OverdraftLimit).
Метод Withdraw() должен быть переопределен для проверки и применения
лимита овердрафта при снятии денег со счета.

3. Инвестиционный счет (InvestmentAccount) (если требуется третий класс):
Должен содержать дополнительные атрибуты, такие как Список активов
(AssetsList). Метод GetInfo() должен быть переопределен для включения
информации о списках активов в описании счета.

```
using System;


public interface IPayment
{
    void ApplyMonthlyFee();
}
public class BankAccount
{
    private string accountnumber;
    public string AccountNumber 
    {
        get { return accountnumber; }
        set { accountnumber = value; }
    }

    private decimal balance;
    public decimal Balance
    {
        get { return balance; }
        set { balance = value; }
    }

    private string accountype;
    public string AccountType
    {
        get { return accountype; }
        set { accountype = value; }
    }

    public string Initials {get; set;}

    private int TransactionsThisMounth = 0;

    public static int TotalAccounts {get; private set;} = 0;

    public string Password {get; set;}

    public string BankName {get; set;}

    
    

    


    public BankAccount(string accountNumber, decimal balance, string accountType, string initials, string bankname, string password) 
    {
        AccountNumber = accountNumber;
        Balance = balance;
        AccountType = accountType;
        Initials = initials;
        BankName = bankname;
        Password = password;
        TotalAccounts++;
    }

    public virtual void GetInfo()
    {
        Console.WriteLine($"Ваш Баланс: {Balance}");
        Console.WriteLine($"Ваш номер аккаунта: {AccountNumber}");
        Console.WriteLine($"Ваш тип аккаунта: {AccountType}");
        Console.WriteLine($"Владельцем карты является: {Initials}");
        Console.WriteLine($"Оформлено в банке: {BankName}");
        Console.WriteLine($"Пароль: {Password}");
    }

    public virtual void Deposit(decimal amount)
    {
        Balance += amount;
        TransactionsThisMounth++;
    }

    public void Deposit(decimal amount, string comment)
    {
        
        Balance += amount;
        TransactionsThisMonth++;
        Console.WriteLine($"Внесено: {amount} руб. ({comment})");
    }

    public virtual void Withdraw(decimal amount)
    {
        Balance -= amount;
        TransactionsThisMounth++;
    }



    public virtual void TransferOnPhoneNumber(decimal amount, string phonenumber)
    {
        Balance -= amount;

        Console.WriteLine($"Вы перевели {amount} рублей на телефон: {phonenumber}");
        TransactionsThisMounth++;
    }

    public virtual void ShowTransactionsThisMounth()
    {
        Console.WriteLine($"Количество операций: {TransactionsThisMounth}");
    }

    public virtual void ChangePassword(string _newPassword)
    {
        Password = _newPassword;

        Console.WriteLine($"Ваш пароль изменен на {_newPassword}");
    }

       public static void ShowTotalAccounts()
    {
        Console.WriteLine($" Всего открыто счетов: {TotalAccounts}");
    }
}




public class SavingsAccount : BankAccount
{
    public decimal InterestRate { get; set; }

    public SavingsAccount(string accountNumber, decimal balance, string accountType, string initials,string bankname, string password, decimal interestRate) : base(accountNumber, balance, accountType, initials, bankname, password)
{
    InterestRate = interestRate;
}

    public override void Deposit(decimal amount)
    {
        decimal interest = Balance * (InterestRate / 100);
        Balance += interest;
        Console.WriteLine($"Начислены проценты: {interest} руб. (ставка {InterestRate}%)");
        Balance += amount;
        Console.WriteLine($"Внесено: {amount} руб. Новый баланс: {Balance} руб.");
    }
}

public class CheckingAccount : BankAccount, IPayment
{
    public decimal OverdraftLimit { get; set; }

    public decimal Monthlypayment { get; set;}

    

   public CheckingAccount(string accountNumber, decimal balance, string accountType, string initials,string bankname, string password, decimal overdraftLimit) : base(accountNumber, balance, accountType, initials, bankname, password)
    {
    OverdraftLimit = overdraftLimit;
    Monthlypayment = 400;
    }

    public override void Withdraw(decimal amount)
    {
        if (amount <= Balance + OverdraftLimit)
        {
            Balance -= amount;
            Console.WriteLine($"Снято: {amount} руб. Новый баланс: {Balance} руб.");
        }
        else
        {
            Console.WriteLine($"Превышен лимит овердрафта!");
        }
    }

        public void ApplyMonthlyFee()
    {
        Balance -= Monthlypayment;
    }
}

public class InvestmentAccount : SavingsAccount
{
    // public string[] AssetsList { get; set; }
    public List<string> AssetsList {get; set;}
    private int _assetsCount;

    public InvestmentAccount(string accountNumber, decimal balance, string accountType, string initials, string bankname, string password, decimal interestRate, List<string> assetsList) : base(accountNumber, balance, accountType, initials, bankname, password, interestRate)
    {
    AssetsList = assetsList ?? new List<string>();
    
     }

    public override void GetInfo()
    {
        base.GetInfo();
        if (AssetsList.Count > 0)
        {
            Console.WriteLine($"Активы: {string.Join(", ", AssetsList)}");
        }
        else
        {
            Console.WriteLine("Активы: нет активов");
        }
    }


    public void AddAssets(string newAsset)
    {
        AssetsList.Add(newAsset);
        Console.WriteLine($"Актив {newAsset} добавлен");
    }

     public void RemoveAsset(string assetToRemove)
    {
        if (AssetsList.Remove(assetToRemove))
        {
            Console.WriteLine($"Актив {assetToRemove} удалён");
        }
        else
        {
            Console.WriteLine($"Актив {assetToRemove} не найден");
        }
    }


}

public class BankCollection<T> where T : BankAccount
{
    private List<T> _accounts = new List<T>();

    public void Add(T account)
    {
        _accounts.Add(account);
        Console.WriteLine($"Счёт {account.AccountNumber} добавлен в коллекцию");
    }

    public void Remove(T account)
    {
        if (_accounts.Remove(account))
            Console.WriteLine($"Счёт {account.AccountNumber} удалён");
        else
            Console.WriteLine($"Счёт не найден");
    }

    public void DisplayAll()
    {
        Console.WriteLine("\n=== Все счета в коллекции ===");
        foreach (var account in _accounts)
        {
            account.GetInfo();
            Console.WriteLine();
        }
    }

    public decimal GetTotalBalance()
    {
        decimal total = 0;
        foreach (var account in _accounts)
            total += account.Balance;
        return total;
    }

    public int Count()
    {
        return _accounts.Count;
    }
}


List<string> myAssets = new List<string> { "Акции Газпром", "Облигации ОФЗ", "Акции Сбер" };

BankAccount myAccount = new BankAccount("ACC-001", 50000, "Обычный", "Гоголев В.М.", "Т-Банк", "95693");
SavingsAccount mysavingaccount = new SavingsAccount("SAV-001", 78932, "Сберегательный", "Гоголев В.М.", "Т-Банк", "95693", 12);
CheckingAccount mycheckingaccount = new CheckingAccount("CHK-001", 20000, "Текущий", "Гоголев В.М.",  "Т-Банк", "95693",  100000);
InvestmentAccount myInvestmentaccount = new InvestmentAccount("INV-001", 30000, "Инвестиционный", "Гоголев В.М.",  "Т-Банк", "95693",  15,  myAssets);

myAccount.GetInfo();
Console.WriteLine();
mysavingaccount.GetInfo();
Console.WriteLine();
mycheckingaccount.GetInfo();
Console.WriteLine();
myInvestmentaccount.GetInfo();
Console.WriteLine();



Console.WriteLine("Обычный счёт:");
myAccount.Deposit(10000);

Console.WriteLine("\nСберегательный счёт (с процентами):");
mysavingaccount.Deposit(5000);

Console.WriteLine("\nИнвестиционный счёт (с процентами):");
myInvestmentaccount.Deposit(3000);

Console.WriteLine();

Console.WriteLine("--- 3. Снятие средств ---\n");

Console.WriteLine("Обычный счёт:");
myAccount.Withdraw(2000);

Console.WriteLine("--- 4. Ежемесячная комиссия ---\n");
Console.WriteLine("Текущий счёт:");
mycheckingaccount.ApplyMonthlyFee();
Console.WriteLine($"Баланс после комиссии: {mycheckingaccount.Balance}");

Console.WriteLine();

Console.WriteLine("--- 5. Перевод на телефон ---\n");

myAccount.TransferOnPhoneNumber(500, "+7-999-123-45-67");
Console.WriteLine($"Баланс после перевода: {myAccount.Balance}");

Console.WriteLine();

Console.WriteLine("--- 6. Добавление активов ---\n");

Console.WriteLine("Инвестиционный счёт до добавления:");
myInvestmentaccount.GetInfo();

Console.WriteLine("\nДобавляем активы:");
myInvestmentaccount.AddAssets("Акции Яндекс");
myInvestmentaccount.AddAssets("Золото");

Console.WriteLine("\nИнвестиционный счёт после добавления:");
myInvestmentaccount.GetInfo();

Console.WriteLine();

Console.WriteLine("--- 7. Статистика операций ---\n");

Console.WriteLine("Обычный счёт:");
myAccount.ShowTransactionsThisMounth();

Console.WriteLine();





```
