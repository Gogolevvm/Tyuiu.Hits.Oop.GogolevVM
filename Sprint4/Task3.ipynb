<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс BankAccount в C#, который будет представлять
информацию об учетных записях в банке. На основе этого класса разработать 2-3
производных класса, демонстрирующих принципы наследования и полиморфизма.
В каждом из классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.


Требования к базовому классу BankAccount:

• Атрибуты: Номер счета (AccountNumber), Баланс (Balance), Тип счета
(AccountType).


• Методы:
o GetInfo(): метод для получения информации о счете в виде строки.
o Deposit(decimal amount): метод для внесения денег на счет.
o Withdraw(decimal amount): метод для снятия денег со счета.

Требования к производным классам:

1. Сберегательный счет (SavingsAccount): Должен содержать дополнительные
атрибуты, такие как Процентная ставка (InterestRate). Метод Deposit() должен
быть переопределен для добавления процентов к сумме вклада при
внесении денег на счет.

2. Текущий счет (CheckingAccount): Должен содержать дополнительные
атрибуты, такие как Лимит овердрафта (OverdraftLimit).
Метод Withdraw() должен быть переопределен для проверки и применения
лимита овердрафта при снятии денег со счета.

3. Инвестиционный счет (InvestmentAccount) (если требуется третий класс):
Должен содержать дополнительные атрибуты, такие как Список активов
(AssetsList). Метод GetInfo() должен быть переопределен для включения
информации о списках активов в описании счета.

```
using System;


public interface IBankAccount
{
    void GetInfo();
    void Deposit(decimal amount);
    void Withdraw(decimal amount);
}

public interface IPayment
{
    void ApplyMonthlyFee();
}
public abstract class BankAccount : IBankAccount
{
 public string AccountNumber { get; set; }
    public decimal Balance { get; set; }
    public string AccountType { get; set; }
    public string Initials { get; set; }
    protected int TransactionsThisMonth = 0;
    

    


    public BankAccount(string accountNumber, decimal balance, string accountType, string initials)
    {
        AccountNumber = accountNumber;
        Balance = balance;
        AccountType = accountType;
        Initials = initials;
        
    }

    public virtual void GetInfo()
    {
        Console.WriteLine($"Ваш Баланс: {Balance}");
        Console.WriteLine($"Ваш номер аккаунта: {AccountNumber}");
        Console.WriteLine($"Ваш тип аккаунта: {AccountType}");
    }

    public virtual void Deposit(decimal amount)
    {
        Balance += amount;
        TransactionsThisMonth++;
    }

    public virtual void Withdraw(decimal amount)
    {
        Balance -= amount;
        TransactionsThisMonth++;
    }



    public virtual void TransferOnPhoneNumber(decimal amount, string phonenumber)
    {
        Balance -= amount;

        Console.WriteLine($"Вы перевели {amount} рублей на телефон: {phonenumber}");
        TransactionsThisMonth++;
    }

    public virtual void ShowTransactionsThisMonth()
    {
        Console.WriteLine($"Количество операций: {TransactionsThisMonth}");
    }


}

public class SavingsAccount : BankAccount
{
    public decimal InterestRate { get; set; }

    public SavingsAccount(string accountNumber, decimal balance, string accountType, string initials, decimal interestRate) : base(accountNumber, balance, accountType, initials)
{
    InterestRate = interestRate;
}

    public override void Deposit(decimal amount)
    {
        decimal interest = Balance * (InterestRate / 100);
        Balance += interest;
        Console.WriteLine($"Начислены проценты: {interest} руб. (ставка {InterestRate}%)");
        Balance += amount;
        Console.WriteLine($"Внесено: {amount} руб. Новый баланс: {Balance} руб.");
    }
}

public class CheckingAccount : BankAccount, IPayment
{
    public decimal OverdraftLimit { get; set; }

    public decimal Monthlypayment { get; set;}

    

   public CheckingAccount(string accountNumber, decimal balance, string accountType, string initials, decimal overdraftLimit) : base(accountNumber, balance, accountType, initials)
    {
    OverdraftLimit = overdraftLimit;
    Monthlypayment = 400;
    }

    public override void Withdraw(decimal amount)
    {
        if (amount <= Balance + OverdraftLimit)
        {
            Balance -= amount;
            Console.WriteLine($"Снято: {amount} руб. Новый баланс: {Balance} руб.");
        }
        else
        {
            Console.WriteLine($"Превышен лимит овердрафта!");
        }
    }

    void IPayment.ApplyMonthlyFee()
    {
        Balance -= Monthlypayment;
        Console.WriteLine($"Ежемесячная комиссия {Monthlypayment} руб. списана. Баланс: {Balance}");
    }
}

public class InvestmentAccount : SavingsAccount
{
    public string[] AssetsList { get; set; }
    private int _assetsCount;

    public InvestmentAccount(string accountNumber, decimal balance, string accountType, string initials, decimal interestRate, string[] assetsList) : base(accountNumber, balance, accountType, initials, interestRate)
    {
    AssetsList = assetsList ?? new string[0];
    _assetsCount = AssetsList.Length;
     }

    public override void GetInfo()
    {
        base.GetInfo();
        if (AssetsList.Length > 0)
        {
            Console.WriteLine($"Активы: {string.Join(", ", AssetsList)}");
        }
        else
        {
            Console.WriteLine("Активы: нет активов");
        }
    }

    public void AddAssets(string newAsset)
    {
        string[] newArray = new string[_assetsCount + 1];

        for (int i = 0; i <  _assetsCount;  i++)
        {
            newArray[i] = AssetsList[i];
        }

        newArray[_assetsCount] = newAsset;

                AssetsList = newArray;
        _assetsCount++;

        Console.WriteLine($"Актив {newAsset} добавлен");


    }
}

public class BankAccountsInfo
{
    private readonly IBankAccount _bankinfo;

    public BankAccountsInfo(IBankAccount bankaccount)
    {
        _bankinfo = bankaccount;
    }

    public void GetAllInfo()
    {
        _bankinfo.GetInfo();
    }
}
string[] myAssets = { "Акции Газпром", "Облигации ОФЗ", "Акции Сбер", "Акции Лукойл" };
SavingsAccount savings = new SavingsAccount("SAV-001", 78932, "Сберегательный", "Гоголев В.М.", 12);
CheckingAccount checking = new CheckingAccount("CHK-001", 20000, "Текущий", "Гоголев В.М.", 100000);
InvestmentAccount investment = new InvestmentAccount("INV-001", 30000, "Инвестиционный", "Гоголев В.М.", 15, myAssets);


BankAccountsInfo info1 = new BankAccountsInfo(savings);
BankAccountsInfo info2 = new BankAccountsInfo(checking);
BankAccountsInfo info3 = new BankAccountsInfo(investment);

info1.GetAllInfo();
Console.WriteLine();
info2.GetAllInfo();
Console.WriteLine();
info3.GetAllInfo();

Console.WriteLine("\n--- Явная реализация IPayment ---");


IPayment payment = checking;
payment.ApplyMonthlyFee(); 
Console.WriteLine();


savings.Deposit(5000);
checking.Withdraw(150000);
investment.AddAssets("Акции Яндекс");
investment.ShowTransactionsThisMonth();
```
